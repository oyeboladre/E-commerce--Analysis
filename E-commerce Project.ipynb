```python
import pandas as pd

# Ensure the tables are visible
orders = all_tables['orders']
customers = all_tables['customers']
order_items = all_tables['order_items']

query_state_category = """
WITH RankedCategories AS (
    SELECT 
        c.customer_state,
        p.product_category_name_english AS category,
        SUM(oi.price) AS Total_Revenue,
        ROW_NUMBER() OVER(PARTITION BY c.customer_state ORDER BY SUM(oi.price) DESC) as rank
    FROM orders o
    JOIN customers c ON o.customer_id = c.customer_id
    JOIN order_items oi ON o.order_id = oi.order_id
    JOIN products_english p ON oi.product_id = p.product_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_state, p.product_category_name_english
)
SELECT 
    customer_state, 
    category AS Top_Category, 
    Total_Revenue
FROM RankedCategories
WHERE rank = 1
ORDER BY Total_Revenue DESC
LIMIT 10;
"""

result_state_category = pysqldf(query_state_category)
print("Top Category per State (Top 10 States by Revenue):")
print(result_state_category)
```

```python
import pandas as pd

# 1. Make sure the tables are visible
orders = all_tables['orders']
order_items = all_tables['order_items']
products = all_tables['products']
order_reviews = all_tables['order_reviews']

# 2. Write the query to get the percentage of each review score per weight category
query_weight_reviews = """
WITH WeightGroups AS (
    SELECT 
        o.order_id,
        r.review_score,
        CASE 
            WHEN p.product_weight_g < 500 THEN '1. Light (< 500g)'
            WHEN p.product_weight_g BETWEEN 500 AND 5000 THEN '2. Medium (500g - 5kg)'
            ELSE '3. Heavy (> 5kg)'
        END AS weight_category
    FROM orders o
    JOIN order_items oi ON o.order_id = oi.order_id
    JOIN products p ON oi.product_id = p.product_id
    JOIN order_reviews r ON o.order_id = r.order_id
    WHERE o.order_status = 'delivered'
)
SELECT 
    weight_category,
    review_score,
    COUNT(order_id) AS total_orders,
    ROUND(COUNT(order_id) * 100.0 / SUM(COUNT(order_id)) OVER (PARTITION BY weight_category), 2) AS percentage
FROM WeightGroups
GROUP BY weight_category, review_score
ORDER BY weight_category, review_score DESC;
"""

result_weight_reviews = pysqldf(query_weight_reviews)
print("Review Score Distribution by Product Weight:")
print(result_weight_reviews)
```

```python
import pandas as pd

# 1. Ensure tables are visible
orders = all_tables['orders']
order_items = all_tables['order_items']
products_english = all_tables['product_category_translation']
products = all_tables['products']

# 2. Merge the translation into a clean products table first
products_clean = pd.merge(products, products_english, on='product_category_name', how='left')

# 3. Run your query against the clean table
query_weight_products = """
WITH RankedProducts AS (
    SELECT 
        CASE 
            WHEN p.product_weight_g < 500 THEN '1. Light (< 500g)'
            WHEN p.product_weight_g BETWEEN 500 AND 5000 THEN '2. Medium (500g - 5kg)'
            ELSE '3. Heavy (> 5kg)'
        END AS weight_category,
        COALESCE(NULLIF(p.product_category_name_english, ''), 'Uncategorized') AS product_category,
        COUNT(DISTINCT o.order_id) AS Total_Orders,
        ROW_NUMBER() OVER(PARTITION BY 
            CASE 
                WHEN p.product_weight_g < 500 THEN '1. Light (< 500g)'
                WHEN p.product_weight_g BETWEEN 500 AND 5000 THEN '2. Medium (500g - 5kg)'
                ELSE '3. Heavy (> 5kg)'
            END 
            ORDER BY COUNT(DISTINCT o.order_id) DESC) as rank
    FROM orders o
    JOIN order_items oi ON o.order_id = oi.order_id
    JOIN products_clean p ON oi.product_id = p.product_id
    WHERE o.order_status = 'delivered'
      AND p.product_category_name_english IS NOT NULL
      AND p.product_category_name_english != ''
    GROUP BY weight_category, p.product_category_name_english
)
SELECT weight_category, product_category, Total_Orders
FROM RankedProducts
WHERE rank <= 5
ORDER BY weight_category, Total_Orders DESC;
"""

# 4. Execute
result_weight_products = pysqldf(query_weight_products)
print(result_weight_products)
```

```python
import pandas as pd

# 1. Ensure tables are visible
orders = all_tables['orders']

# 2. Write the query
query_late = """
SELECT 
    strftime('%Y-%m', o.order_purchase_timestamp) AS month,
    COUNT(*) AS Total_Delivered_Orders,
    SUM(CASE WHEN o.order_delivered_customer_date > o.order_estimated_delivery_date THEN 1 ELSE 0 END) AS Late_Orders,
    ROUND(
        SUM(CASE WHEN o.order_delivered_customer_date > o.order_estimated_delivery_date THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 
        2
    ) AS Late_Percentage
FROM orders o
WHERE o.order_status = 'delivered'
  AND o.order_delivered_customer_date IS NOT NULL
  AND o.order_estimated_delivery_date IS NOT NULL
GROUP BY month
ORDER BY month;
"""

# 3. Execute
result_late = pysqldf(query_late)

# 4. Print
print("Late Delivery Rate Over Time:")
print(result_late)
```

```python
#Delivery Performance9
import pandas as pd
import plotly.express as px

# 1. Reshape the data so On-Time and Late are separate rows for plotting
result_late['On_Time_Orders'] = result_late['Total_Delivered_Orders'] - result_late['Late_Orders']

# Melt the dataframe to get a 'Delivery Status' column
result_melted = result_late.melt(
    id_vars=['month'],
    value_vars=['On_Time_Orders', 'Late_Orders'],
    var_name='Delivery_Status',
    value_name='Total_Orders'
)

# 2. Create the 100% Stacked Bar Chart
fig_delivery = px.bar(
    result_melted,
    x='month',
    y='Total_Orders',
    color='Delivery_Status',
    title='Delivery Performance: On-Time vs Late Over Time',
    barmode='stack',
    width=1000,
    height=600,
    color_discrete_map={
        'On_Time_Orders': '#4A90E2', 
        'Late_Orders': '#D0021B'      
    }
)

# 3. Clean up the hover
fig_delivery.update_traces(
    hovertemplate="<b>Month: %{x}</b><br>Status: %{fullData.name}<br>Orders: %{y:,.0f}<extra> </extra>"
)

# 4. Update layout
fig_delivery.update_layout(
    xaxis_title="Month",
    yaxis_title="Total Number of Orders",
    legend_title="Delivery Status",
    plot_bgcolor='white'
)

fig_delivery.update_yaxes(showgrid=True, gridwidth=1, gridcolor='LightGray')
fig_delivery.show()
```

```python
import pandas as pd

# 1. Ensure tables are visible
orders = all_tables['orders']
customers = all_tables['customers']
order_items = all_tables['order_items']

# 2. Write the query using customer_unique_id (the real person)
query_loyalty = """
WITH CustomerSpend AS (
    SELECT 
        c.customer_unique_id,
        COUNT(DISTINCT o.order_id) AS Total_Orders,
        SUM(oi.price) AS Total_Spent
    FROM orders o
    JOIN customers c ON o.customer_id = c.customer_id
    JOIN order_items oi ON o.order_id = oi.order_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_unique_id
)
SELECT 
    CASE 
        WHEN Total_Orders = 1 THEN '1. One-Time Buyer'
        ELSE '2. Repeat Buyer'
    END AS customer_type,
    COUNT(customer_unique_id) AS Total_Customers,
    ROUND(AVG(Total_Spent), 2) AS Avg_Lifetime_Spend,
    ROUND(AVG(Total_Orders), 2) AS Avg_Orders_Per_Customer
FROM CustomerSpend
GROUP BY customer_type
ORDER BY customer_type;
"""

result_loyalty = pysqldf(query_loyalty)
print("Customer Loyalty Breakdown:")
print(result_loyalty)
```

```python
import pandas as pd

# 1. Calculate the core metrics using SQL
kpi_query = """
SELECT 
    COUNT(DISTINCT o.order_id) AS total_orders,
    ROUND(SUM(oi.price), 2) AS total_revenue,
    ROUND(AVG(r.review_score), 2) AS avg_review_score,
    ROUND(AVG(oi.freight_value), 2) AS avg_freight_cost,
    ROUND(SUM(CASE WHEN o.order_delivered_customer_date > o.order_estimated_delivery_date THEN 1 ELSE 0 END) * 100.0 / COUNT(DISTINCT o.order_id), 2) AS late_delivery_rate
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
JOIN order_reviews r ON o.order_id = r.order_id
WHERE o.order_status = 'delivered';
"""
kpis = pysqldf(kpi_query)

total_orders = kpis['total_orders'].iloc[0]
total_revenue = kpis['total_revenue'].iloc[0]
avg_review = kpis['avg_review_score'].iloc[0]
avg_freight = kpis['avg_freight_cost'].iloc[0]
late_rate = kpis['late_delivery_rate'].iloc[0]

# 2. Calculate Average Order Value (AOV)
aov = total_revenue / total_orders

# 3. Calculate Repeat Buyer Rate (The complex one!)
loyalty_query = """
WITH CustomerSpend AS (
    SELECT 
        c.customer_unique_id,
        COUNT(DISTINCT o.order_id) AS Total_Orders
    FROM orders o
    JOIN customers c ON o.customer_id = c.customer_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_unique_id
)
SELECT 
    COUNT(customer_unique_id) AS total_customers,
    SUM(CASE WHEN Total_Orders > 1 THEN 1 ELSE 0 END) AS repeat_customers
FROM CustomerSpend;
"""
loyalty_data = pysqldf(loyalty_query)
repeat_rate = (loyalty_data['repeat_customers'].iloc[0] / loyalty_data['total_customers'].iloc[0]) * 100

# 4. Print them beautifully
print("="*40)
print("       EXECUTIVE KPIs DASHBOARD")
print("="*40)
print(f"💰 Total Revenue:      R$ {total_revenue:,.2f}")
print(f"📦 Total Orders:       {total_orders:,}")
print(f"🛒 Avg Order Value:    R$ {aov:,.2f}")
print(f"🚚 Late Delivery Rate: {late_rate:.2f}%")
print(f"⭐ Avg Review Score:   {avg_review} / 5")
print(f"🔄 Repeat Buyer Rate:  {repeat_rate:.2f}%")
print(f"🚛 Avg Freight Cost:   R$ {avg_freight:,.2f}")
print("="*40)
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# 1. Create a 2x4 grid for the KPI cards
fig_kpi = make_subplots(
    rows=2, cols=4,
    specs=[[{"type": "domain"}, {"type": "domain"}, {"type": "domain"}, {"type": "domain"}],
           [{"type": "domain"}, {"type": "domain"}, {"type": "domain"}, {"type": "domain"}]],
    subplot_titles=("Total Revenue", "Total Orders", "Avg Order Value (AOV)", "Late Delivery Rate",
                    "Avg Review Score", "Repeat Buyer Rate", "Avg Freight Cost", "Customer Base")
)

# 2. Add the KPI indicators
# Row 1, Col 1: Total Revenue
fig_kpi.add_trace(go.Indicator(
    mode="number", value=total_revenue, number={"prefix": "R$ ", "valueformat": ",.0f"},
    title={"text": "Total Revenue", "font": {"size": 16, "color": "#00E5FF"}}), row=1, col=1)

# Row 1, Col 2: Total Orders
fig_kpi.add_trace(go.Indicator(
    mode="number", value=total_orders, number={"valueformat": ","},
    title={"text": "Total Orders", "font": {"size": 16, "color": "#00FF88"}}), row=1, col=2)

# Row 1, Col 3: Avg Order Value
fig_kpi.add_trace(go.Indicator(
    mode="number", value=aov, number={"prefix": "R$ ", "valueformat": ",.2f"},
    title={"text": "Avg Order Value", "font": {"size": 16, "color": "#FFB300"}}), row=1, col=3)

# Row 1, Col 4: Late Delivery Rate (Red because it's a warning metric)
fig_kpi.add_trace(go.Indicator(
    mode="number", value=late_rate, number={"suffix": "%", "valueformat": ".2f"},
    title={"text": "Late Delivery Rate", "font": {"size": 16, "color": "#FF0055"}}), row=1, col=4)

# Row 2, Col 1: Avg Review Score
fig_kpi.add_trace(go.Indicator(
    mode="number", value=avg_review, number={"suffix": " / 5"},
    title={"text": "Avg Review Score", "font": {"size": 16, "color": "#FFFFFF"}}), row=2, col=1)

# Row 2, Col 2: Repeat Buyer Rate (Red because it's low)
fig_kpi.add_trace(go.Indicator(
    mode="number", value=repeat_rate, number={"suffix": "%", "valueformat": ".2f"},
    title={"text": "Repeat Buyer Rate", "font": {"size": 16, "color": "#FF0055"}}), row=2, col=2)

# Row 2, Col 3: Avg Freight Cost
fig_kpi.add_trace(go.Indicator(
    mode="number", value=avg_freight, number={"prefix": "R$ ", "valueformat": ",.2f"},
    title={"text": "Avg Freight Cost", "font": {"size": 16, "color": "#FFFFFF"}}), row=2, col=3)

# Row 2, Col 4: Total Unique Customers
fig_kpi.add_trace(go.Indicator(
    mode="number", value=loyalty_data['total_customers'].iloc[0], number={"valueformat": ","},
    title={"text": "Unique Customers", "font": {"size": 16, "color": "#FFFFFF"}}), row=2, col=4)

# 3. Apply Dark Mode Styling
fig_kpi.update_layout(
    title_text="<b>Olist E-Commerce Executive KPIs</b>",
    title_x=0.5,
    title_font=dict(size=26, color='#FFFFFF'),
    height=500,
    width=1200,
    paper_bgcolor='#121212',
    plot_bgcolor='#121212',
    font=dict(color='#FFFFFF')
)

# Save and show
fig_kpi.write_html("kpi_dashboard.html")
fig_kpi.show()
print("✅ KPI Board saved as 'kpi_dashboard.html'")
```
